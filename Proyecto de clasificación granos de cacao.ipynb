```python
import numpy as np
import cv2
import pandas as pd

anchos_procesados = []
altos_procesados = []


def caracteristicas_hsv_rgb(ruta_imagen):
    imagen = cv2.imread(ruta_imagen)

    if imagen is None:
      return pd.Series({
          "hue_promedio" : np.nan,
          "sat_promedio" : np.nan,
          "val_promedio" : np.nan,
          "%_cafe" : np.nan,
          "%_morado" : np.nan,
          "%_oscuro": np.nan,
          "proporcion_llenado" : np.nan,
      })

    IMG_WIDTH = 30
    IMG_HEIGHT = 30
    imagen = cv2.resize(imagen, (IMG_WIDTH, IMG_HEIGHT))

    # CAPTURAMOS LAS DIMENSIONES EXACTAS DESPUÉS DEL RESIZE
    h, w = imagen.shape[:2]
    altos_procesados.append(h)
    anchos_procesados.append(w)


    hsv = cv2.cvtColor(imagen, cv2.COLOR_BGR2HSV)
    saturacion = hsv[:,:,1]
    brillo = hsv[:,:,2]
    mask = (saturacion > 50) & (brillo < 210)

    pixeles = hsv[mask]
    tono = pixeles[:,0]

    rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    R = rgb[:,:,0].astype(float)[mask]
    G = rgb[:,:,1].astype(float)[mask]
    B = rgb[:,:,2].astype(float)[mask]

    es_cafe = (R > G) & (G > B) & ((R-B) > 15)
    es_morado = (R > G) & (B > G) & (np.abs(R-B) < 45)
    es_oscuro = brillo[mask] < 40

    area_grano = np.sum(mask)
    area_total = imagen.shape[0] * imagen.shape[1]
    proporcion_llenado = area_grano / area_total

    return pd.Series({
        "hue_promedio" : np.mean(tono),
        "sat_promedio" : np.mean(saturacion[mask]),
        "val_promedio" : np.mean(brillo[mask]),
        "%_cafe" : np.mean(es_cafe)*100,
        "%_morado" : np.mean(es_morado*100),
        "%_oscuro": np.mean(es_oscuro) * 100,
        "proporcion_llenado" : proporcion_llenado,
    })

df_limpio[["hue_promedio", "sat_promedio", "val_promedio", "%_cafe", "%_morado","%_oscuro", "proporcion_llenado"]] = df_limpio["Ruta"].apply(caracteristicas_hsv_rgb)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(anchos_procesados, bins=30, color='skyblue')
axes[0].axvline(np.median(anchos_procesados), color="red", linestyle="--", label=f"Mediana: {int(np.median(anchos_procesados))}")
axes[0].set_title("Ancho de las fotos procesadas")
axes[0].legend()

axes[1].hist(altos_procesados, bins=30, color='salmon')
axes[1].axvline(np.median(altos_procesados), color="red", linestyle="--", label=f"Mediana: {int(np.median(altos_procesados))}")
axes[1].set_title("Alto de las fotos procesadas")
axes[1].legend()

plt.show()

print("Ancho exacto verificado:", int(np.median(anchos_procesados)))
print("Alto exacto verificado:", int(np.median(altos_procesados)))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, confusion_matrix, ConfusionMatrixDisplay,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

FEATURES = ["sat_promedio", "val_promedio",
            "%_morado","%_oscuro", "proporcion_llenado" ]
COLUMNA_CLASE = "Clase"
ORDEN = ["Bueno", "Parcial", "Malo"]

X = df_limpio[FEATURES]
y = df_limpio["Clase"].map({c: i for i, c in enumerate(ORDEN)})

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

def metricas(y_real, y_pred, titulo):
    print("Confusion Matrix:\n", confusion_matrix(y_real, y_pred))
    ConfusionMatrixDisplay(confusion_matrix(y_real, y_pred), display_labels=ORDEN).plot(cmap="Blues")
    plt.title("Matriz de confusión - " + titulo)
    plt.show()

    cm = confusion_matrix(y_real, y_pred)
    print("==================================================")
    print("     MÉTRICAS INDIVIDUALES POR CLASE")
    print("==================================================")

    specificities = []

    for i, clase in enumerate(ORDEN):
        VP = cm[i, i]
        FP = cm[:, i].sum() - VP
        FN = cm[i, :].sum() - VP
        VN = cm.sum() - (VP + FP + FN)

        precision = VP / (VP + FP) if (VP + FP) > 0 else 0.0
        recall = VP / (VP + FN) if (VP + FN) > 0 else 0.0
        especificidad = VN / (VN + FP) if (VN + FP) > 0 else 0.0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
        accuracy_clase = (VP + VN) / (VP + VN + FP + FN) if (VP + VN + FP + FN) > 0 else 0.0

        specificities.append(especificidad)

        print(f"\n--- CLASE: {clase} ---")
        print(f"  • Precisión:      {precision:.6f}")
        print(f"  • Sensibilidad:   {recall:.6f}")
        print(f"  • Especificidad:  {especificidad:.6f}")
        print(f"  • Accuracy:       {accuracy_clase:.6f}")
        print(f"  • F1-Score:       {f1:.6f}")

    especificidad_general = sum(specificities) / len(specificities)

    print("\n==================================================")
    print("     MÉTRICAS GENERALES")
    print("==================================================")
    print(f"Accuracy Score: {accuracy_score(y_real, y_pred):.6f}")
    print(f"Sensibilidad:   {recall_score(y_real, y_pred, average='macro', zero_division=0):.6f}")
    print(f"Precision:      {precision_score(y_real, y_pred, average='macro', zero_division=0):.6f}")
    print(f"F1 Score:       {f1_score(y_real, y_pred, average='macro', zero_division=0):.6f}")
    print(f"Especificidad:  {especificidad_general:.6f}")
    print("\n")
```

```python
import cv2
import numpy as np
import tensorflow as tf

IMG_HEIGHT = 30
IMG_WIDTH = 30

X_img = []
for ruta in df_limpio["Ruta"]:
    img = cv2.imread(ruta)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_WIDTH, IMG_HEIGHT), interpolation=cv2.INTER_AREA)
    X_img.append(img)

X_img = np.array(X_img).astype("float32")
y_img = y.values

x_train, x_test, y_train, y_test = train_test_split(
    X_img,
    y_img,
    test_size=0.3,
    random_state=42,
    stratify=y_img
)

print(x_train.shape, y_train.shape, x_test.shape, y_test.shape)
print(tf.__version__)
```

```python
x_train = x_train.reshape([-1, 30, 30, 3])
x_test = x_test.reshape([-1, 30, 30, 3])

x_train = x_train.astype('float32')
x_test = x_test.astype('float32')

x_train /= 255.0
x_test/= 255.0

print('x_train shape:', x_train.shape)
print('Number of images in x_train:', x_train.shape[0])
print('Number of images in x_test:', x_test.shape[0])
```

```python
import tensorflow as tf
from tensorflow import keras

model1 = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=[30, 30, 3]),
    tf.keras.layers.Dense(1024, activation=tf.nn.relu),
    tf.keras.layers.Dense(512, activation=tf.nn.relu),
    tf.keras.layers.Dense(256, activation=tf.nn.relu),
    tf.keras.layers.Dense(128, activation=tf.nn.relu),
    tf.keras.layers.Dense(64, activation=tf.nn.relu),
    tf.keras.layers.Dense(3, activation=tf.nn.softmax)
])

model1.summary()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
from scipy.cluster import hierarchy
from scipy.spatial.distance import cdist

sns.set_theme()
X = df_limpio[FEATURES]
X = StandardScaler().fit_transform(X)
y = df_limpio["Clase"].map({c: i for i, c in enumerate(ORDEN)}).values

print("Granos:", X.shape[0], "| Características:", X.shape[1])
```

```python
#@title **KMEANS**
distortions = []
wss = []
K = range(1, 11)
for k in K:
    kmeanModel = KMeans(n_clusters=k, n_init='auto', random_state=42).fit(X)
    distortions.append(sum(np.min(cdist(X, kmeanModel.cluster_centers_, 'euclidean'), axis=1)) / X.shape[0])
    wss.append(kmeanModel.inertia_)

plt.figure(figsize=(12, 4))
plt.subplot(121)
plt.plot(K, distortions, 'bx-'); plt.xlabel('k'); plt.ylabel('Distortion')
plt.title('El método del codo (distorsión)')
plt.subplot(122)
plt.plot(K, wss, 'bx-'); plt.xlabel('k'); plt.ylabel('Inercia')
plt.title('El método del codo (inercia)')
plt.show()
```
