```python
import cv2
import numpy as np
import pandas as pd


TAM = (128, 128)

kernel = np.array([[ 0, -1,  0],
                   [-1,  5, -1],
                   [ 0, -1,  0]])

lista_matrices, indices_validos = [], []

for i, ruta in df["Ruta"].items():
    img = cv2.imread(ruta)
    if img is None:
        print(f"No se pudo leer: {ruta}")
        continue
    img = cv2.resize(img, TAM, interpolation=cv2.INTER_AREA)
    if kernel is not None:
        img = cv2.filter2D(img, -1, kernel)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
    lista_matrices.append(img.flatten())      # cada imagen -> un vector fila
    indices_validos.append(i)

# DataFrame: una fila por grano, una columna por píxel/canal
pix_cols = [f"p{j}" for j in range(len(lista_matrices[0]))]
df_pix = pd.DataFrame(np.array(lista_matrices), columns=pix_cols)

# Se agregan las etiquetas y metadatos del df original
meta = df.loc[indices_validos, ["Escena", "Semilla", "Clase"]].reset_index(drop=True)
df_pix = pd.concat([meta, df_pix], axis=1)

ORDEN = ["Bueno", "Parcial", "Malo"]
df_pix["y"] = df_pix["Clase"].map({c: i for i, c in enumerate(ORDEN)})

print(df_pix.shape)
print(df_pix["Clase"].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Partición train/test (estratificada por el desbalance de clases)
X = df_pix[pix_cols]
y = df_pix["y"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

# 2. Escalado (ajustado solo con train)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Reducción dimensional con PCA (ajustada solo con train)
pca = PCA(n_components=0.50, random_state=42)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

print(f"Componentes: {pca.n_components_} (de {X_train.shape[1]} columnas originales)")

FEATURES = [f"PC{i+1}" for i in range(pca.n_components_)]
```

### **Logistic Regressor**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay, accuracy_score,
                             recall_score, precision_score, f1_score)

modelR = LogisticRegression(class_weight="balanced", max_iter=1000)
modelR.fit(X_train_pca, y_train)

y_pred = modelR.predict(X_test_pca)

print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred), display_labels=ORDEN).plot(cmap="Blues")
plt.title("Matriz de confusión - Regresión Logística (PCA)")
plt.show()

print("Accuracy Score:", accuracy_score(y_test, y_pred))
print("Sensibilidad:", recall_score(y_test, y_pred, average="weighted"))
print("Precision:", precision_score(y_test, y_pred, average="weighted"))
print("F1 Score:", f1_score(y_test, y_pred, average="weighted"))
print("\n")

# Especificidad por clase a partir de la matriz de confusión
cm = confusion_matrix(y_test, y_pred)
especificidad = []

for i in range(len(cm)):
    VP = cm[i, i]
    FN = cm[i, :].sum() - VP
    FP = cm[:, i].sum() - VP
    VN = cm.sum() - (VP + FN + FP)

    esp = VN / (VN + FP) if (VN + FP) > 0 else 0
    especificidad.append(esp)

for clase, valor in zip(ORDEN, especificidad):
    print(f"{clase}: {valor:.4f}")

print("Especificidad promedio:", sum(especificidad) / len(especificidad))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
from scipy.cluster import hierarchy
from scipy.spatial.distance import cdist

sns.set_theme()

# Sin train/test: no hay entrenamiento supervisado, se agrupan todos los granos
N_PC = 10    # número de componentes principales (el k del PCA)
FEATURES2 = [f"PC{i+1}" for i in range(N_PC)]

pix_scaled = StandardScaler().fit_transform(df_pix[pix_cols])      # píxeles estandarizados
X = PCA(n_components=N_PC, random_state=42).fit_transform(pix_scaled)
y = df_pix["y"].values                                              # 0=Malo, 1=Parcial, 2=Bueno

print("Granos:", X.shape[0], "| Características:", X.shape[1])
```

```python
#@title **KMEANS**
distortions = []
wss = []
K = range(1, 11)
for k in K:
    kmeanModel = KMeans(n_clusters=k, n_init='auto', random_state=42).fit(X)
    distortions.append(sum(np.min(cdist(X, kmeanModel.cluster_centers_, 'euclidean'), axis=1)) / X.shape[0])
    wss.append(kmeanModel.inertia_)

plt.figure(figsize=(12, 4))
plt.subplot(121)
plt.plot(K, distortions, 'bx-'); plt.xlabel('k'); plt.ylabel('Distortion')
plt.title('El método del codo (distorsión)')
plt.subplot(122)
plt.plot(K, wss, 'bx-'); plt.xlabel('k'); plt.ylabel('Inercia')
plt.title('El método del codo (inercia)')
plt.show()
```

```python
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay, accuracy_score,
                             recall_score, precision_score, f1_score)

def metricas(y_real, y_pred, titulo):
    etiquetas = list(range(len(ORDEN)))              # [0, 1, 2] -> Malo, Parcial, Bueno
    cm = confusion_matrix(y_real, y_pred, labels=etiquetas)

    print("Confusion Matrix:\n", cm)
    ConfusionMatrixDisplay(cm, display_labels=ORDEN).plot(cmap="Blues")
    plt.title("Matriz de confusión - " + titulo)
    plt.show()

    print("==================================================")
    print("     MÉTRICAS INDIVIDUALES POR CLASE")
    print("==================================================")

    specificities = []

    for i, clase in enumerate(ORDEN):
        VP = cm[i, i]
        FP = cm[:, i].sum() - VP
        FN = cm[i, :].sum() - VP
        VN = cm.sum() - (VP + FP + FN)

        precision = VP / (VP + FP) if (VP + FP) > 0 else 0.0
        recall = VP / (VP + FN) if (VP + FN) > 0 else 0.0
        especificidad = VN / (VN + FP) if (VN + FP) > 0 else 0.0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
        accuracy_clase = (VP + VN) / (VP + VN + FP + FN) if (VP + VN + FP + FN) > 0 else 0.0

        specificities.append(especificidad)

        print(f"\n--- CLASE: {clase} ---")
        print(f"  • Precisión:      {precision:.6f}")
        print(f"  • Sensibilidad:   {recall:.6f}")
        print(f"  • Especificidad:  {especificidad:.6f}")
        print(f"  • Accuracy:       {accuracy_clase:.6f}")
        print(f"  • F1-Score:       {f1:.6f}")

    especificidad_general = sum(specificities) / len(specificities)

    print("\n==================================================")
    print("     MÉTRICAS GENERALES")
    print("==================================================")
    print(f"Accuracy Score: {accuracy_score(y_real, y_pred):.6f}")
    print(f"Sensibilidad:   {recall_score(y_real, y_pred, average='macro', labels=etiquetas, zero_division=0):.6f}")
    print(f"Precision:      {precision_score(y_real, y_pred, average='macro', labels=etiquetas, zero_division=0):.6f}")
    print(f"F1 Score:       {f1_score(y_real, y_pred, average='macro', labels=etiquetas, zero_division=0):.6f}")
    print(f"Especificidad:  {especificidad_general:.6f}")
    print("\n")
```

```python
for k_val in [2, 3]:
    print(f"\n==================================================")
    print(f"          EVALUACIÓN PARA K = {k_val}")
    print(f"==================================================")

    # Ajustar KMeans
    km = KMeans(n_clusters=k_val, n_init='auto', random_state=42)
    km.fit(X)

    # Mapeo de clusters a la clase mayoritaria real
    mapa_k = {}
    for cluster_id in range(k_val):
        clase_mayoritaria = np.bincount(y[km.labels_ == cluster_id]).argmax()
        mapa_k[cluster_id] = clase_mayoritaria

    # Predicciones basadas en la clase mayoritaria
    y_pred_km_k = np.array([mapa_k[c] for c in km.labels_])

    # Mostrar métricas
    metricas(y, y_pred_km_k, f"K-means (K={k_val})")
```
